```python
import pandas as pd
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
path = '/content/drive/MyDrive/praktikum 11/data/UFO_sightings_Spain.csv'
```

```python
df = pd.read_csv(path)
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
X = df[['longitude', 'latitude']]
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import DBSCAN
dbscan = DBSCAN(eps=0.2, min_samples=5)
df['Cluster_Label'] = dbscan.fit_predict(X_scaled)
```

```python
df['Cluster_Label_Str'] = df['Cluster_Label'].astype(str)
```

```python
fig = px.scatter_mapbox(
    df,
    lat="latitude",
    lon="longitude",
    color="Cluster_Label_Str",  # Gunakan label klaster sebagai warna
    hover_data={'longitude': True, 'latitude': True, 'Cluster_Label': True},
    zoom=4.5, # Tingkat zoom agar Spanyol terlihat jelas
    height=600,
    mapbox_style="carto-positron", # Menggunakan peta dasar yang ringan
    title="Peta Klaster Penampakan UFO di Spanyol (DBSCAN)"
)
```

```python
fig.update_traces(marker=dict(size=10, opacity=0.8))
fig.update_layout(
    legend_title_text='Klaster',
    margin={"r":0,"t":50,"l":0,"b":0}
)
```
